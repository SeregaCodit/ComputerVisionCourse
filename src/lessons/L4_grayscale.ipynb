```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

from core.entities.constants import Paths as p
```

### GrayscalePixel = 0.299 R + 0.587 G + 0.114 B

```python
IMAGE_PATH = p.IMAGE_ASSETS / "cat1.png"
```

```python
img_bgr = cv2.imread(IMAGE_PATH)

plt.figure()
plt.imshow(img_bgr)
img_bgr.shape
```

```python
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
plt.figure()
plt.imshow(img_gray, cmap="gray") # by default matplotlib uses viridis colormap
img_gray.shape
```
