```python
import cv2

import numpy as np
import matplotlib.pyplot as plt

from core.entities.constants import Paths as p
```

```python
img_path = p.IMAGE_ASSETS / "cat1.png"
img = cv2.imread(img_path)

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure()
plt.imshow(img_rgb)
```

```python
height, width, _ = img_rgb.shape
angle = 180

scale = 1
center = (int(width / 2), int(height / 2))

cv2.circle(img_rgb, center, 5, (0, 0, 255), -1)

T = cv2.getRotationMatrix2D(center, angle, scale)
img_rotated = cv2.warpAffine(img_rgb, T, (width, height))
cv2.putText(img_rotated,
            f"{center}",
            (center[0] + 10, center[1] - 10),
            cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)

plt.figure()
plt.imshow(img_rotated)
```
