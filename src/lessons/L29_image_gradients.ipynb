```python
import cv2 as cv
import matplotlib.pyplot as plt

from core.entities.constants import Paths as p
```

```python
img_path = p.IMAGE_ASSETS / "lambo.jpg"
img = cv.imread(img_path)

img = cv.imread(img_path, cv.IMREAD_GRAYSCALE)

plt.figure()
plt.imshow(img, cmap='gray')
```

```python
laplacian = cv.Laplacian(img, cv.CV_64F, ksize=3)
plt.figure()
plt.imshow(laplacian, cmap='gray')
```

```python
kx, ky = cv.getDerivKernels(1, 0, 3)
kernel = ky@kx.T
print(f"kx:\n{kx}", end="\n\n")
print(f"ky:\n{ky}", end="\n\n")
print(f"kernel:\n{kernel}")
```

```python
sobel_x = cv.Sobel(img, cv.CV_64F, 1, 0, ksize=3)

plt.figure()
plt.imshow(sobel_x, cmap='gray')
```

```python
sobel_y = cv.Sobel(img, cv.CV_64F, 0, 1, ksize=3)
plt.figure()
plt.imshow(sobel_y, cmap='gray')
```

```python
sobel_xy = cv.addWeighted(sobel_x, 0.5, sobel_y, 0.5, 0)
plt.figure()
plt.imshow(sobel_xy, cmap='gray')
```
