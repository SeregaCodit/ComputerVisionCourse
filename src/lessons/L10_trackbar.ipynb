```python
import cv2
import matplotlib.pyplot as plt

from core.entities.constants import Paths as p
```

```python
IMG_PATH = p.IMAGE_ASSETS / "cat1.png"
img = cv2.imread(IMG_PATH)

plt.figure()
plt.imshow(img)
```

```python
def trackbar_callback(value):
    pass
```

```python
widow_name = "trackbar"
cv2.namedWindow(widow_name)
cv2.createTrackbar("B", widow_name, 0, 255, trackbar_callback)
cv2.createTrackbar("G", widow_name, 0, 255, trackbar_callback)
cv2.createTrackbar("R", widow_name, 0, 255, trackbar_callback)
```

```python
while True:
    cv2.imshow(widow_name, img)

    if cv2.waitKey(1) == ord("q"):
        break

    b = cv2.getTrackbarPos("B", widow_name)
    g = cv2.getTrackbarPos("G", widow_name)
    r = cv2.getTrackbarPos("R", widow_name)

    cv2.circle(img, (466, 251), 10, (b, g, r), -1)
    cv2.circle(img, (566, 238), 10, (b, g, r), -1)

cv2.destroyAllWindows()
```
