```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

from core.entities.constants import Paths as p
from infrastructure.utils.plotting import subplot
```

```python
image_path = p.IMAGE_ASSETS / "cat1.png"
```

```python
img = cv2.imread(image_path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.figure()
plt.imshow(img)
```

```python
img_crop = img[225:275, 434:484, :]
plt.figure()
plt.imshow(img_crop)
```

```python
interp_methods = [
    cv2.INTER_AREA,
    cv2.INTER_CUBIC,
    cv2.INTER_LINEAR,
    cv2.INTER_LANCZOS4,
    cv2.INTER_NEAREST,
]

interp_title = ["area", "cubic", "linear", "lanczos", "nearest"]
```

```python
def resize_and_plot_image(img: np.ndarray, scale: float) -> None:
    height, width, _ = img.shape
    plt.figure()
    plt.subplot(2, 3, 1)
    plt.title("original")
    plt.imshow(img)

    for i in range (len(interp_methods)):
        method = interp_methods[i]
        title = interp_title[i]
        plt.subplot(2, 3, i + 2)
        img_resize = cv2.resize(img, (int(width * scale), int(height * scale)), interpolation=method)

        plt.title(title)
        plt.imshow(img_resize)
```

```python
resize_and_plot_image(img_crop, 0.25)
```

```python
resize_and_plot_image(img_crop, 4)
```
