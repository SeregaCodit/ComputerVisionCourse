```python
def avg_filter_callback(value):
    pass
```

```python
window_name = "avg_filter"
cv2.namedWindow(window_name)

cv2.createTrackbar("k_size", window_name, 1, 100, avg_filter_callback)
```

```python
while True:
    if cv2.waitKey(1) == ord("q"):
        break

    k_size = cv2.getTrackbarPos("k_size", window_name)
    k_size = max(k_size, 1)

    img_blured = cv2.blur(img, (k_size, k_size))

    cv2.imshow(window_name, img_blured)

cv2.destroyAllWindows()
```
