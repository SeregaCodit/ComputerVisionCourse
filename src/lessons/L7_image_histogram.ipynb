```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

from core.entities.constants import Paths as p
```

```python
IMAGE_PATH = p.IMAGE_ASSETS / "cat1.png"
image = cv2.imread(IMAGE_PATH, cv2.IMREAD_COLOR_RGB)

plt.figure()
plt.imshow(image)
```

# Gray Histogram

```python
grayscale = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

plt.figure()
plt.imshow(grayscale, cmap="gray")
```

```python
hist = cv2.calcHist([grayscale], [0], None, [256], [0, 256])
plt.figure()
plt.plot(hist)
plt.xlabel("color intense")
plt.ylabel("# of Pixels")
```

# Color Histogram

```python
def plot_colors_hist(img: np.ndarray, colors: list[str,]):

    for i in range(len(colors)):
        hist = cv2.calcHist([img], [i], None, [256], [0, 256])
        plt.plot(hist, colors[i])

    plt.xlabel("color intense")
    plt.ylabel("# of Pixels")
```

```python
colors = ["b", "g", "r"]
```

```python
plot_colors_hist(image, colors)
```

# A Region Hist

```python
face_crop = image[137:372, 356:676, :]

plt.figure()
plt.imshow(face_crop)
```

```python
plot_colors_hist(face_crop, colors)
```
