```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

from core.entities.constants import Paths as p
from infrastructure.utils.plotting import subplot
```

### PURE COLORS

```python
SIZE = (100, 100)

# by default np returns matrices in np.float64 dtype
zeros = np.zeros(SIZE)
ones = np.ones(SIZE)

plt.figure()
plt.subplot(121)
plt.title("zeros")
zeros_img = cv2.merge([zeros, zeros, zeros])  # same cv2.merge([zeros] * 3)
plt.imshow(zeros_img)

plt.subplot(122)
plt.title("ones")
ones_img = cv2.merge([ones] * 3) # same cv2.merge([ones, ones, ones])
plt.imshow(ones_img)


```

```python
plt.figure()
blue = cv2.merge([zeros, zeros, 255 * ones]).astype(np.uint8)
green = cv2.merge([ones, 255 * ones, zeros]).astype(np.uint8)
red = cv2.merge([255 * ones, zeros, zeros]).astype(np.uint8)

plt.subplot(131)
plt.title("blue")
plt.imshow(blue)

plt.subplot(132)
plt.title("green")
plt.imshow(green)

plt.subplot(133)
plt.title("red")
plt.imshow(red)
```

### BGR Grayscale

```python
IMAGE_PATH = p.IMAGE_ASSETS / "cat1.png"
```

```python
img_bgr = cv2.imread(IMAGE_PATH)
plt.figure()
plt.imshow(img_bgr)
```

```python
b , g, r = cv2.split(img_bgr)

data = dict(
    blue = b,
    green = g,
    red = r,
)
```

```python
subplot(data)
```

```python
subplot(data, cmap = "gray")
```

```python
# create a channel with B channel's dimensions
zeros_channel = np.zeros_like(b)
print(f"zeros_channel.shape: {zeros_channel.shape}\n"
      f"B_channel.shape: {b.shape}\n")
subplot({"zeros": zeros_channel})
```

```python
b_image = cv2.merge([zeros_channel, zeros_channel, b])
g_image = cv2.merge([zeros_channel, g, zeros_channel])
r_image = cv2.merge([r, zeros_channel, zeros_channel])

data = dict(
    blue = b_image,
    green = g_image,
    red = r_image,
)

subplot(data)
```
