```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

from core.entities.constants import Paths as p
```

```python
IMAGE_PATH = p.IMAGE_ASSETS / "cat1.png"

img = cv2.imread(IMAGE_PATH, cv2.IMREAD_COLOR_RGB)

plt.figure()
plt.imshow(img)
```

```python
GREEN_COLOR = (175, 255, 175)
```

```python
# line
cv2.line(img, (477, 293), (322, 409), GREEN_COLOR, 2)

plt.figure()
plt.imshow(img)
```

```python
# rectangle
r, c, d = img.shape
offset = 10

cv2.rectangle(img, (offset, offset), (c - offset, r - offset), GREEN_COLOR, 4)

plt.figure()
plt.imshow(img)
```

```python
# circle

cv2.circle(img, (557, 239), 15, GREEN_COLOR, 2)
cv2.circle(img, (454, 250), 15, GREEN_COLOR, -1)
plt.figure()
plt.imshow(img)
```

```python
cv2.ellipse(img, (500, 276), (30, 15), 0, 0, 180, GREEN_COLOR, -1)
plt.figure()
plt.imshow(img)
```

```python
pts = np.array([[588, 138],
                [633, 109],
                [656, 138],
                [644, 205]])

cv2.polylines(img, [pts], True, GREEN_COLOR, 2)
plt.figure()
plt.imshow(img)
```

```python
cv2.putText(img, "WHAT A TIGER!", (42, 776), cv2.FONT_HERSHEY_SIMPLEX, 2, GREEN_COLOR, 2)
plt.figure()
plt.imshow(img)
```
