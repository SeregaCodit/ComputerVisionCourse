    cv2.bilateralFilter(
        src,
        d,
        sigmaColor,
        sigmaSpace
    )

**d** - діаметр просторового neighborhood. Фільтр аналізує локальну область навколо кожного пікселя. Kоли d має позитивне значення, він визначає діаметр околу; при d <= 0 розмір визначається через sigmaSpace

**sigmaColor** - визначає, наскільки сильно фільтр буде дозволяти змішувати пікселі з різними значеннями кольору/інтенсивності.

**sigmaSpace** - наскільки далеко від центрального пікселя можуть ефективно впливати сусіди.


| Властивість | Gaussian | Bilateral |
|---|---:|---:|
| Зменшує шум | ✅ | ✅ |
| Згладжує | ✅ | ✅ |
| Враховує відстань | ✅ | ✅ |
| Враховує різницю інтенсивності | ❌ | ✅ |
| Зберігає edges | ❌ | ✅ |
| Лінійний | ✅ | ❌ |
| Швидкість | Вища | Нижча |
| Обчислювальна складність | Нижча | Вища |
