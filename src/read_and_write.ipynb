### WEB_CAM


```
def video_from_webcam():
    cap = cv2.VideoCapture(1)

    if not cap.isOpened():
        exit()


    while True:
        ret, frame = cap.read()

        if ret:
            cv2.imshow("webcam", frame)

            if cv2.waitKey(1) & 0xFF == ord('q'):
                break

    cap.release()
    cv2.destroyAllWindows()
```

```
VIDEO_ASSETS = Path(os.getcwd()).parent / "assets/videos"
VIDEO_ASSETS
```

```
def video_from_file(file_path: Path):
    cap = cv2.VideoCapture(file_path)
    fps = cap.get(cv2.CAP_PROP_FPS)
    delay = int(1000 / fps)

    if not cap.isOpened():
        raise ValueError(f"Could not read video from file: {file_path}")

    while cap.isOpened():
        ret, frame = cap.read()

        if not ret:
            break

        cv2.imshow(file_path.name, frame)

        if cv2.waitKey(delay) & 0xFF == ord('q'):
            break

    cap.release()
    cv2.destroyAllWindows()
```

```
def write_video_file(outputpath: Path):
    cap = cv2.VideoCapture(1)

    fourcc = cv2.VideoWriter_fourcc(*"XVID")
    out = cv2.VideoWriter(str(outputpath), fourcc, 20.0, (640, 480))

    while cap.isOpened():
        ret, frame = cap.read()

        if ret:
            out.write(frame)
            cv2.imshow("webcam", frame)

            if cv2.waitKey(1) & 0xFF == ord('q'):
                break

    cap.release()
    out.release()
    cv2.destroyAllWindows()
```

```
filename = "cat.mp4"
filepath = VIDEO_ASSETS / filename
```

```
video_from_file(filepath)
```

```
output_file = "saved_cat.avi"
output_path = VIDEO_ASSETS / output_file
```

```
write_video_file(output_path)
```
